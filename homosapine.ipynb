```python
from Bio import SeqIO, Entrez, Align


# simple understanding
Entrez.email = "your@email.com"



def get_cds(acc):
  '''
  acc is a string  species number eg: NM_000518.5 (homo sapiens)..

  '''

  data = Entrez.efetch(db="nucleotide", id=acc, rettype="fasta", retmode="text")
  rec = SeqIO.read(data, "fasta")

  seq = rec.seq
  start = seq.find("ATG") # the gene starts at ATG, first start codon
  protein = seq[start: ].translate(to_stop = True) # amino acids    # to_stop = True is a parameter or argument      in protein only contain amino seqence not the utr
  cds = seq[start:start + 3 * len(protein)]           # get the DNA letters that made the protein


  return cds, protein

  # help(rec.seq)


human_cds, human_prot  = get_cds("NM_000518")  # return tuple







print("Human CDs", len(human_cds), "Codon | amino_acid", len(human_prot),"amino_acid")
print("Human Amino_acid starts:", human_prot[:20])


```
